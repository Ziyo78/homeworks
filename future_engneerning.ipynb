```python
ohe_title_status = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

ohe_title_status.fit(df_1[['title_status']])
is_title_status = pd.DataFrame(ohe_title_status.transform(df_1[['title_status']]), columns = 'is_' + ohe_title_status.get_feature_names_out())
```

```python
ohe_transmission = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

ohe_transmission.fit(df_1[['transmission']])
is_transmission = pd.DataFrame(ohe_transmission.transform(df_1[['transmission']]), columns = 'is_' + ohe_transmission.get_feature_names_out())
```

```python
df_1 =  pd.concat([df_1, is_transmission], axis=1)
df_1 = df_1.drop(columns=['transmission'])
```
